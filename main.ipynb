### Chunk initial dataset in batches of 425,000 observations and save them as separate CSV files.

```python
def chunk_dataframe(chunk_size, output_dir, base_file_path, header=0):
    """
    Splits a DataFrame into smaller chunks and saves them as CSV files.

    Parameters:
    chunk_size (int): The number of rows per chunk.
    output_dir (str): The directory where the chunked CSV files will be saved.
    base_file_path (str): The path to the base CSV file.
    header (int or list of int): Row number(s) to use as the column names.

    Returns:
    None
    """
    with pd.read_csv(base_file_path, header=header, chunksize=chunk_size) as reader:
        for i, chunk in enumerate(reader):
            chunk.to_csv(f"{output_dir}/{os.path.basename(base_file_path).split('.')[0]}_chunk_{i + 1}.csv", index=False)
            print(f"Saved chunk {i + 1} to {output_dir}/{os.path.basename(base_file_path).split('.')[0]}_chunk_{i + 1}.csv")
```

### Majority class classifier implementation

```python
class MajorityClassClassifier:
    """
       Majority Class Classifier that predicts the majority class for each label based on the training data.
    """
    
    def __init__(self):
        self.N_t = 0  # Total number of training samples
        self.occ_1 = pd.DataFrame()  # Total counts of class 1 by each column (label) in the training data
        self.predicted_probabilities = pd.DataFrame()  # Predicted probabilities for each label

    def fit(self, Y):
        """
        Fit the classifier to the training data.

        Parameters:
        X (pd.DataFrame): The training data.
        Y (pd.DataFrame): The labels for the training data.

        Returns:
        None
        """
        self.N_t += Y.shape[0]  # Update the total number of training samples
        # Add to the total counts of class 1 by each column (label) in the training data
        if self.occ_1.empty:
            self.occ_1 = Y.sum()  # Initialize with the sum of occurrences of class 1 across all labels
        else:
            self.occ_1 += Y.sum()  # Sum of all occurrences of class 1 across all labels

    def predict(self, N_t):
        """
        Predict the labels for the given data.

        Parameters:
        N_t (int): The number of training samples.
        
        Returns:
        pd.DataFrame: The predicted probabilities for each label.
        """
        if self.N_t == 0:
            raise ValueError("The classifier has not been fitted yet. Please call the 'fit' method first.")
        # Calculate the predicted probabilities for each label
        self.predicted_probabilities = self.occ_1 / self.N_t
        # Create a DataFrame with the predicted probabilities for each label, repeated for each sample in X
        return pd.DataFrame(np.tile(self.predicted_probabilities.values, (N_t, 1)), columns=self.predicted_probabilities.index)

   
```

### LogLoss Implementation

```python
#Log loss tradeshift text classification
def log_loss(Y_true, Y_pred):
    """
    Computes the log loss for multi-class classification.

    Parameters:
    Y_true (pd.DataFrame): True labels
    Y_pred (pd.DataFrame): Predicted probabilities
    
    Returns:
    float: Log loss value
    """
    # Ensure Y_true and Y_pred have the same shape
    if Y_true.shape != Y_pred.shape:
        raise ValueError("Y_true and Y_pred must have the same shape.")
    
    N_t = Y_true.shape[0]
    K = Y_true.shape[1]

    # Clip predicted probabilities to avoid log(0)
    Y_pred = np.clip(Y_pred, 1e-15, 1 - 1e-15)

    return np.sum(-(Y_true * np.log(Y_pred)) - ((1 - Y_true) * np.log(1 - Y_pred))) / (N_t * K)
```

### Sparsing

```python
## Online Learning and sparsity implementation from scratch

# Parse training data
def parse_x(train_data):
    """
    Parses the training data and returns a sparse representation of the features.

    Parameters:
    train_data (pd.DataFrame): The training data.

    Returns:
    np.ndarray: A sparse representation of the features.
    """
    # Initialize an empty list to store the sparse representation of the features
    x_sparse = [np.array([]) for _ in range(train_data.shape[0])]

    # Iterate over each row in the training data
    for index, row in train_data.iterrows():
        non_zero_elements = np.zeros(DIMENSION, dtype=int)
        for col_name, value in row.items():
            if value != 0 and not pd.isna(value):
                non_zero_elements[hash(col_name+'_'+str(value)) % DIMENSION] = 1
        x_sparse[index] = non_zero_elements
    return x_sparse
```

```python
# # estimate belongship probability at each labels using the current weight vector
# def predict_proba(x_sparse, w):
#     """
#     Estimates the probability of belonging to a class for a given input using the current weight vector.

#     Parameters:
#     x_sparse (np.ndarray): The sparse representation of the input features.
#     w (np.ndarray): The current weight vector.

#     Returns:
#     float: The estimated probability of belonging to a class.
#     """
#     # Calculate the dot product of the sparse input and the weight vector
#     print(f"x_sparse shape: {x_sparse.shape}")
#     print(f"x_sparse values: {x_sparse}")

#     print(f"w shape: {w.T.shape}")
#     print(f"w values: {w.T}")
#     z = np.dot(x_sparse, w.T)
    
#     print(z)
#     # Apply the sigmoid function to get the probability
#     probability = 1 / (1 + np.exp(-z))

#     print(f"Predicted probability shape: {probability.shape}")
    
#     return probability 

# estimate belongship probability at each label using the current weight matrix
def predict_proba(x_sparse, W):
    """
    Estimates the probability of belonging to each class for a given input using the current weight matrix.

    Parameters:
    x_sparse (np.ndarray): The sparse representation of the input features, shape (DIMENSION,).
    W (np.ndarray): The current weight matrix, shape (K, DIMENSION).

    Returns:
    np.ndarray: The estimated probability of belonging to each label, shape (K,).
    """
    # Calculate the dot product of the sparse input and the weight matrix -> one score per label
    z = np.dot(W, x_sparse)  # shape (K,)

    # Apply the sigmoid function to get the probability for each label
    probability = 1 / (1 + np.exp(-z))

    return probability
```

```python
# # Update the weight vector using  gradient descent
# def update_weights(w, grd, x_sparse, y_true, alpha=0.01, beta=1.0):
#     """
#     Updates the weight vector using gradient descent.

#     Parameters:
#     w (np.ndarray): The current weight vector.
#     x_sparse (np.ndarray): The sparse representation of the input features.
#     y_true (int): The true label for the input.
#     alpha (float): The regularization parameter.
#     beta (float): The learning rate for the update.

#     Returns:
#     np.ndarray: The updated weight vector.
#     """
#     # Estimate the probability of belonging to the class
#     y_pred = predict_proba(x_sparse, w)
#     print(f"Predicted probabilities shape: {y_pred.shape}, True label shape: {y_true.shape}")
#     grd += (abs(y_true - y_pred))** 2 # square of the gradient of the loss function with respect to the weights

#     n = alpha / (beta + np.sqrt(grd))

#     w -= beta * n  # update the weights using gradient descent

#     return w

# %%
# Update the weight matrix using gradient descent (per-label, AdaGrad-style)
def update_weights(W, grd, x_sparse, y_true, alpha=0.01, beta=1.0):
    """
    Updates the weight matrix using gradient descent, one row per label.

    Parameters:
    W (np.ndarray): The current weight matrix, shape (K, DIMENSION).
    grd (np.ndarray): The accumulated squared gradients, shape (K, DIMENSION).
    x_sparse (np.ndarray): The sparse representation of the input features, shape (DIMENSION,).
    y_true (np.ndarray): The true labels for the input, shape (K,).
    alpha (float): The regularization parameter.
    beta (float): The learning rate for the update.

    Returns:
    tuple: (updated W, updated grd)
    """
    # Estimate the probability of belonging to each class
    y_pred = predict_proba(x_sparse, W)  # shape (K,)

    # Per-label error, broadcast against each label's feature row
    error = (y_true - y_pred)  # shape (K,)

    # Gradient for each label w.r.t. its own weight row: error_k * x_sparse
    grad = np.outer(error, x_sparse)  # shape (K, DIMENSION)

    grd += grad ** 2  # accumulate squared gradient per label, per feature

    n = alpha / (beta + np.sqrt(grd))  # shape (K, DIMENSION)

    W += beta * n * grad  # gradient ASCENT on log-likelihood (note sign fix below)

    return W, grd
```

```python
def parse_y(train_label_data):
    """
    Parses the labels from the training data.

    Parameters:
    train_label_data (pd.DataFrame): The training label data.

    Returns:
    np.ndarray: The parsed labels.
    """
    return train_label_data.iloc[:, 1:].values
```

```python
# # Train the model using online learning
# def train_online(train_data, train_label_data, learning_rate=0.01):
#     """
#     Trains the model using online learning.

#     Parameters:
#     train_data (pd.DataFrame): The training data.
#     train_label_data (pd.DataFrame): The training label data.
#     learning_rate (float): The learning rate for the update.

#     Returns:
#     np.ndarray: The trained weight vector.
#     """
#     # Initialize weights
#     w = np.zeros(DIMENSION)

#     # Initialize the gradient accumulator
#     grd = np.zeros(DIMENSION)
    
#     # Parse training data and labels
#     x_sparse = parse_x(train_data)
#     y_true = parse_y(train_label_data)
    
#     # Iterate over each sample in the training data
#     for i in range(len(x_sparse)):
    #     # Update weights using the current sample
    #     w = update_weights(w, grd, x_sparse[i], y_true[i], beta=learning_rate)
    
    # return w
# Train the model using online learning
def train_online(train_data, train_label_data, learning_rate=0.01):
    """
    Trains the model using online learning.

    Parameters:
    train_data (pd.DataFrame): The training data.
    train_label_data (pd.DataFrame): The training label data.
    learning_rate (float): The learning rate for the update.

    Returns:
    np.ndarray: The trained weight matrix, shape (K, DIMENSION).
    """
    # Parse training data and labels first, so we know K
    x_sparse = parse_x(train_data)
    y_true = parse_y(train_label_data)
    K = y_true.shape[1]

    # Initialize weights: one row per label
    W = np.zeros((K, DIMENSION))

    # Initialize the gradient accumulator: same shape as W
    grd = np.zeros((K, DIMENSION))

    # Iterate over each sample in the training data
    for i in range(len(x_sparse)):
        # Update weights using the current sample
        W, grd = update_weights(W, grd, x_sparse[i], y_true[i], beta=learning_rate)

    return W
```
